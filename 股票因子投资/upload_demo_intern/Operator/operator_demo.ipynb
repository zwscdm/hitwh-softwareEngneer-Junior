```python
from Operator import *

# 实例化
operator = Operator()

start_day = '2019-01-01'
end_day = '2023-12-19'
```

```python
# 日数据算子：有roll_前缀
return_daily = pd.read_pickle('/daily/ret.pkl').loc[start_day:end_day]
return_daily
```

```python
# 计算20日收益峰度
operator.roll_kurt(factor=return_daily,window=20)
# 计算20日收益最大值
operator.roll_max(factor=return_daily,window=20)

close_daily = pd.read_pickle('/daily/close.pkl').loc[start_day:end_day]
# 计算20日收盘价最大回撤率
operator.max_downup_rate_day(factor=close_daily,window=20,drawdown=True)

industry_l1 = pd.read_pickle('/daily/l1_indus.pkl').bfill().ffill() # 行业数据若存在缺失值，需要填充缺失值
# 计算一级行业分类横截面均值
operator.sample_classify_mean(factor=return_daily, classify=industry_l1)
# 剔除一级行业风格影响
operator.factor_sample_classify(factor=return_daily, classify_list=[industry_l1], corr_period=20, method="mean")
```

```python
# min数据算子：无roll_前缀
return_5min = pd.read_pickle('/5min/ret.pkl').loc[start_day:end_day]
return_5min
```

```python
# 计算5min数据日内时序排名
operator.tsrank_min(factor=return_5min,pct=True)
# 计算1日5min峰度
operator.kurt(factor=return_5min)
# 计算1日5min最大值
operator.max(factor=return_5min)

turnover_5min = pd.read_pickle('/5min/turnover_float.pkl').loc[start_day:end_day]
# 计算一日13:05-15:00，5min换手率均值
operator.part_mean(factor=turnover_5min, begin_time='13:05', end_time='15:00')

close_5min = pd.read_pickle('/5min/close.pkl').loc[start_day:end_day]
# 计算1日5min收盘价最大涨幅
operator.max_downup_rate_min(factor=close_5min, drawdown=False)
```

```python
# 将daily数据repeat至 1min or 5min数据
return_daily = pd.read_pickle('/daily/ret.pkl').loc[start_day:end_day]
return_5min =  pd.read_pickle('/5min/ret.pkl').loc[start_day:end_day]
index_5min = return_5min.index

operator.repeat_min(factor=return_daily, min_date_name=index_5min, begin_time='09:35', end_time='15:00')
```
