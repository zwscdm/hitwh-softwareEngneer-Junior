```python
import numpy as np
import pandas as pd

data_path = '../basic_data'
```

### 日频数据

```python
# 读取数据
return_daily = pd.read_pickle(f'{data_path}/daily/ret.pkl')
return_daily
```

```python
# 日频基础算子函数
def roll_skew(factor, window):
    return factor.rolling(window,min_periods=3).skew()

def roll_wmean(factor, weight, window):
    return (factor*weight).rolling(window,min_periods=1).sum() / weight.rolling(window,min_periods=1).sum()

def roll_mean(factor, window):
    return factor.rolling(window,min_periods=2).mean()
```

```python
# 日频因子函数
def factor_daily_001(return_daily):
    ret_skew = roll_skew(return_daily, 10)
    ret_mean = roll_mean(return_daily, 10)
    factor = roll_wmean(ret_mean, ret_skew , 10)
    return factor

factor1 = factor_daily_001(return_daily=return_daily)
factor1
```

### min频数据

```python
# 读取数据
turnover_5min = pd.read_pickle(f'{data_path}/5min/turnover_float.pkl')
turnover_5min
```

```python
# min频基础算子
def skew(factor):
        return_df = factor.groupby(factor.index.date).skew()
        return_df.index = pd.to_datetime(return_df.index).strftime("%Y-%m-%d")
        return return_df

def mean(factor):
        return_df = factor.groupby(factor.index.date).mean()
        return_df.index = pd.to_datetime(return_df.index).strftime("%Y-%m-%d")
        return return_df
```

```python
# min因子算子
def factor_5min_001(turnover_5min):
    ret_skew = skew(turnover_5min)
    ret_mean = mean(turnover_5min)
    factor = roll_wmean(ret_mean, ret_skew , 10)
    return factor

factor2 = factor_5min_001(turnover_5min=turnover_5min)
factor2
```
